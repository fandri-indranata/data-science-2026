```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

# Langkah 1: Muat Dataset Telco Customer Churn (menggunakan raw URL publik yang stabil)
# Trying another public URL for the dataset
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("=== Shape Data ===")
print(df.shape)
print("\n=== Proporsi Kelas Churn ===")
print(df["Churn"].value_counts(normalize=True).round(3))
```

```python
# Langkah 2: Preprocessing
# 1. Ubah target 'Churn' menjadi numerik (Yes=1, No=0)
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# 2. Drop kolom 'customerID' karena tidak relevan untuk prediksi
df = df.drop('customerID', axis=1)

# 3. One-Hot Encoding untuk semua fitur kategorikal
df_encoded = pd.get_dummies(df, drop_first=True)

# 4. Pisahkan Fitur (X) dan Target (y)
X = df_encoded.drop('Churn', axis=1)
y = df_encoded['Churn']

# 5. Train-Test Split dengan stratify untuk menjaga proporsi kelas
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print(f"Data Latih: {X_tr.shape}, Data Uji: {X_te.shape}")
```
